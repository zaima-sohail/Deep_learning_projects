```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("salader/dogsvscats")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
import tensorflow
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from keras.layers import Dense,Flatten
from keras.applications.vgg16 import VGG16
```

```python
convo_base=VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(150,150,3)
)
```

```python
model=Sequential()
model.add(VGG16(include_top=False,input_shape=(224,224,3)))
model.add(Flatten())
model.add(Dense(256,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
convo_base.trainable=False
```

```python
train_ds=tensorflow.keras.utils.image_dataset_from_directory(
    directory=path,
    labels='inferred',
    label_mode='int',
    batch_size=32,
    image_size=(224,224)
)
```

```python
val_ds=tensorflow.keras.utils.image_dataset_from_directory(
    directory=path,
    labels='inferred',
    label_mode='int',
    batch_size=32,
    image_size=(224,224))
```

```python
def process(image,label):
    image=tensorflow.cast(image/255.0,tensorflow.float32)
    return image,label
train_ds=train_ds.map(process)
val_ds=val_ds.map(process)
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
histroy=model.fit(train_ds,epochs=10,validation_data=val_ds)
```
