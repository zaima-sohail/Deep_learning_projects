```python
def initalize_parameters(layers_dim):
  np.random.seed(3)
  parameters={}
  L=len(layers_dim)
  for l in range(1,L):
    parameters['w'+str(l)] = np.ones((layers_dim[l-1], layers_dim[l])) * 0.1
    parameters['b'+str(l)] = np.zeros((layers_dim[l], 1))
  return parameters
```

```python
initalize_parameters([2,2,1])
```

```python
def linear_forward(A_prev,w,b):
  y_pred=np.dot(w.T,A_prev)+b
  return y_pred
```

```python
def forward_propragation(X,parameters):
   A=X
   L=len(parameters)//2
   A1_output = None # To store the activation of the first layer

   for l in range(1,L+1):
    A_prev=A
    w1=parameters['w'+str(l)]
    b1=parameters['b'+str(l)]

    print("A"+str(l-1)+": ", A_prev)
    print("W"+str(l)+": ", w1)
    print("b"+str(l)+": ", b1)

    A = linear_forward(A_prev, w1, b1)

    if l == 1:
        A1_output = A # Store A after the first layer

    print("A"+str(l)+": ", A)
    print("-"*20)
   return A, A1_output # Return both the final activation and A1
```

```python
def update_parameters(parameters, y, y_hat, A1, X):

    # Update for W2
    parameters['w2'][0][0] = parameters['w2'][0][0] + \
        (0.001 * 2 * (y - y_hat) * A1[0][0])

    parameters['w2'][1][0] = parameters['w2'][1][0] + \
        (0.001 * 2 * (y - y_hat) * A1[1][0])

    # Update for b2
    parameters['b2'][0][0] = parameters['b2'][0][0] + \
        (0.001 * 2 * (y - y_hat))


    # Update for W1 (first component)
    parameters['w1'][0][0] = parameters['w1'][0][0] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][0][0] * X[0][0])

    parameters['w1'][1][0] = parameters['w1'][1][0] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][0][0] * X[1][0])

    # Update for b1 (first component)
    parameters['b1'][0][0] = parameters['b1'][0][0] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][0][0])

    # Update for W1 (second component)
    parameters['w1'][0][1] = parameters['w1'][0][1] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][1][0] * X[0][0])

    parameters['w1'][1][1] = parameters['w1'][1][1] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][1][0] * X[1][0])

    # Update for b1 (second component)
    parameters['b1'][1][0] = parameters['b1'][1][0] + \
        (0.001 * 2 * (y - y_hat) * parameters['w2'][1][0])
```
