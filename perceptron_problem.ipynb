```python
from sklearn.linear_model import Perceptron
clf1=Perceptron()
clf2=Perceptron()
clf3=Perceptron()
```

```python
clf1.fit(and_data.iloc[:,0:2], and_data.iloc[:,-1].values)
```

```python
clf2.fit(or_data.iloc[:,0:2], or_data.iloc[:,-1].values)
```

```python
clf3.fit(xor_data.iloc[:,0:2].values, xor_data.iloc[:,-1].values)
```

```python
X=np.linspace(-1,1,5)
y=-X+1
```

```python
plt.plot(X,y)
sns.scatterplot(x=or_data['input1'], y=or_data['input2'], hue=or_data['output'], s=100)
```

```python
clf3.coef_
clf3.intercept_
```

```python
def plot_decision_region(X, y):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(
        np.arange(x_min, x_max, 0.01),
        np.arange(y_min, y_max, 0.01)
    )
    grid = np.c_[xx1.ravel(), xx2.ravel()]
    predictions = clf3.predict(grid).reshape(xx1.shape)
    plt.contourf(xx1, xx2, predictions, alpha=0.3, cmap='coolwarm')
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='black', s=100)
    plt.xlim(x_min, x_max)
    plt.ylim(y_min, y_max)
    plt.xlabel('input1')
    plt.ylabel('input2')
    plt.show()
```

```python
plot_decision_region(xor_data.iloc[:,0:2].values,xor_data.iloc[:,-1].values)
```
