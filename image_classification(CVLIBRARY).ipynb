```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image=cv2.imread("flower.jpg")
if image is None:
  print("inage not found!")
  exit()
  cv2.imshow(" originial image",image)

```

```python
from google.colab.patches import cv2_imshow

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
from google.colab.patches import cv2_imshow

# Display the original image
cv2_imshow(image)
```

```python
blur=cv2.GaussianBlur(gray,(5,5),0)
cv2_imshow(blur)
```

```python
median=cv2.medianBlur(gray,5)
cv2_imshow(median)
```

```python
# Get image dimensions (height, width)
h, w = image.shape[:2]

# 1. Rotation: Rotate by 45 degrees around the center
center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))

# 2. Scaling: Resize the image to 50% of its original size
scaled_image = cv2.resize(image, (None, None), fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

# 3. Translation: Shift the image x=+50, y=+30
translation_matrix = np.float32([[1, 0, 50], [0, 1, 30]])
translated_image = cv2.warpAffine(image, translation_matrix, (w, h))

# Display the results
print("--- Rotated Image (45 Degrees) ---")
cv2_imshow(rotated_image)

print("\n--- Scaled Image (50% Size) ---")
cv2_imshow(scaled_image)

print("\n--- Translated Image (Shifted Right & Down) ---")
cv2_imshow(translated_image)
```

```python
# 1. Sobel Edge Detection (Combining X and Y gradients)
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_combined = cv2.magnitude(sobel_x, sobel_y)
# Convert back to an 8-bit unsigned integer image
sobel_final = np.uint8(np.clip(sobel_combined, 0, 255))

# 2. Canny Edge Detection (Applying thresholds of 100 and 200)
canny_edges = cv2.Canny(gray, 100, 200)

# Display the results
print("--- Sobel Edge Detection ---")
cv2_imshow(sobel_final)

print("\n--- Canny Edge Detection ---")
cv2_imshow(canny_edges)
```
